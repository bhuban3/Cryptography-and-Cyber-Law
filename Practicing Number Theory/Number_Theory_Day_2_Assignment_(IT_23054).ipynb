# Problem 1: Repunit Divisibility

```python
def repunit_mod(n, p):
    r = 0

    for i in range(n):
        r = (r * 10 + 1) % p

    return r


def find_Ap(p, limit=100):
    result = []

    for n in range(1, limit + 1):
        if repunit_mod(n, p) == 0:
            result.append(n)

    return result


primes = [2, 3, 5, 7, 11, 13, 17, 19, 23]

for p in primes:
    print(f"Prime p = {p}")
    print("A_p =", find_Ap(p))
    print("-" * 40)
```

# Problem 2: Even Perfect Numbers

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True


def find_even_perfect_numbers(limit):
    for n in range(2, limit + 1):
        mersenne = 2**n - 1

        if is_prime(mersenne):
            perfect = 2**(n-1) * mersenne

            print("n =", n)
            print("Mersenne number =", mersenne)
            print("Perfect number =", perfect)
            print("-" * 40)


find_even_perfect_numbers(20)
```

# Verify using the sum of proper divisors

```python
def is_perfect(num):
    divisor_sum = 0

    for i in range(1, num):
        if num % i == 0:
            divisor_sum += i

    return divisor_sum == num


numbers = [6, 28, 496, 8128]

for num in numbers:
    print(num, "is perfect:", is_perfect(num))
```

# Problem 3: Fermat Pseudoprimes


```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False

    return True


def generate_pseudoprime(a, p):

    if not is_prime(p):
        return None

    if p == 2 or (a * (a**2 - 1)) % p == 0:
        return None

    n = (a**(2*p) - 1) // (a**2 - 1)

    return n


bases = [2, 3, 4, 5]
primes = [3, 5, 7, 11, 13, 17, 19, 23]

for a in bases:

    print("\nBase =", a)

    for p in primes:

        n = generate_pseudoprime(a, p)

        if n is not None:

            composite = not is_prime(n)
            fermat = pow(a, n-1, n) == 1

            print("p =", p)
            print("Pseudoprime candidate =", n)
            print("Composite =", composite)
            print("Fermat condition =", fermat)
            print("-" * 40)
```
